# Problem & Objective defination
### Problem
- The dataset describes 6476 rows of up-to-date information on tourist expenditure collected by the National Bureau of Statistics (NBS) in Tanzania.The dataset was collected to gain a better understanding of the status of the tourism sector and provide an instrument that will enable sector growth.

### Objective
- To accurately predict tourist expenditure when visiting Tanzania.

### More info
- The majority of the visitors under the age group of 25-44 came for business (18.5%), or leisure and holidays (53.2%), which is consistent with the fact that they are economically more productive. Those at the age group of 45-64 were more prominent in holiday making and visiting friends and relatives. The results further reveal that most visitors belonging to the age group of 18-24 came for leisure and holidays (55.3%) as well as volunteering (13.7%). The majority of senior citizens (65 and above) came for leisure and holidays (80.9%) and visiting friends and relatives (9.5%).

- The survey covers seven departure points, namely: Julius Nyerere International Airport, Kilimanjaro International Airport, Abeid Amani Karume International Airport, and the Namanga, Tunduma, Mtukula and Manyovu border points.

### Points
- This is a regression problem
- The target variable is expenditure ```total_cost```
- Evaluation will be MEA(Mean Absolute Error)

### Data Cleaning

````python
train_data.isnull().sum()
````

````python
## Fill the categirical columns with mode
fill_cols = ['travel_with', 'most_impressing']
for col in fill_cols:
    mode_val = train_data[col].mode()[0]
    train_data[col] = train_data[col].fillna(mode_val)
````

````python
## Fill remaining null with bfill and ffill
train_data = train_data.ffill()
train_data = train_data.bfill()
````

````python
train_data.isnull().sum()
````

````python
train_data.duplicated().sum()
````

````python
# Check for outliers
num_cols = ['total_cost', 'total_female', 'total_male', 'night_mainland', 'night_zanzibar']

def check_outliers(cols):
    for col in cols:
        sns.boxplot(data=train_data[col])
        plt.title(f'Boxplot of {col}')
        plt.show()

check_outliers(num_cols)
````

````python
# Remove outlier using IQR methodl
print('Old shape: ', train_data.shape)

def remove_outlier(train_data):
    num_cols = ['total_cost', 'total_female', 'total_male', 'night_mainland', 'night_zanzibar']
    for col in num_cols:
        Q1 = train_data[col].quantile(0.25)
        Q3 = train_data[col].quantile(0.75)
        IQR = Q3-Q1
        lower = Q1 - IQR*1.5
        upper = Q3 + IQR*1.5

        train_data = train_data[(train_data[col] >= lower) & (train_data[col] <= upper)]
    return train_data

train_data = remove_outlier(train_data)
print('New shape: ', train_data.shape)
````

````python
# Check for outliers
check_outliers(num_cols)
````

## Grouping into high and low 

````python
# 1. age_group
train_data['age_group_split'] = train_data['age_group'].apply(
    lambda x: 'high' if x in ['65+', '45-64'] else 'low'
)
````

````python
# 2. travel_with split
train_data['travel_with_split'] = train_data['travel_with'].apply(
    lambda x: 'high' if x in ['Spouse', 'Spouse and Children'] else 'low'
)
````

````python
# 3.purpose split
train_data['purpose_split'] = train_data['purpose'].apply(
    lambda x: 'high' if x in ['Leisure and Holidays', 'Volunteering'] else 'low'
)
````

````python
# 4. main_activity split
train_data['main_activity_split'] = train_data['main_activity'].apply(
    lambda x: 'high' if x in ['business', 'Beach tourism', 'Wildlife tourism', 'Conference tourism', 'Diving and Sport Fishing'] else 'low'
)
````

````python
# 5. info_source split
train_data['info_source_split'] = train_data['info_source'].apply(
    lambda x: 'high' if x in ['Travel, agent, tour operator', 'Radio, Tv, Web', 'Newspaper, magazines,brochures'] else 'low'
)
````

````python
# 6. Creating package score
package_cols = ['tour_arrangement', 'package_transport_int', 'package_accomodation', 'package_food', 'package_transport_tz',
                'package_sightseeing', 'package_guided_tour', 'package_insurance']

# Convert tour_arrangemnt to Yes or No for encoding
train_data['tour_arrangement'] = train_data['tour_arrangement'].apply(
    lambda x: 'Yes' if x == 'Package Tour' else 'No'
)

# Encode the values first: 1 -> Yes, 0 -> No
for col in package_cols:
    train_data[col] = train_data[col].map({'Yes': 1, 'No': 0})

# Calculate package score
train_data['package_score'] = train_data[package_cols].sum(axis=1)
````

````python
# 7. Encode first_trip_tz
train_data['first_trip_tz'] = train_data['first_trip_tz'].apply(
    lambda x: 1 if x == 'Yes' else 0
)
````

````python
# 8. most_impressing split
train_data['most_impressing_split'] = train_data['most_impressing'].apply(
    lambda x: 'high' if x in ['Wildlife', 'Good Service', 'Wonderful Country, Landscape, Nature', 'Excellent Experience'] else 'low'
)
````

````python
# 9. Split payment_mode
train_data['payment_mode_split'] = train_data['payment_mode'].apply(
    lambda x: 'high' if x in ['Other', 'Credit Card'] else 'low'
)
````

````python
# 10. Split country
country_cols = ['LUXEMBOURG', 'PHILIPINES', 'RUSSIA', 'SINGAPORE', 'SRI LANKA', 'MALT'
                'ARGENTINA', 'COLOMBIA', 'HUNGARY', 'CROATIA', 'TAIWAN', 'NAMIBIA', 'ROMANIA', 'CROATIA']

train_data['country_split'] = train_data['country'].apply(
    lambda x: 'high' if x in country_cols else 'low'
)
````

````python
# 11. Encode for the split columns
split_cols = ['travel_with_split', 'purpose_split', 'main_activity_split', 'info_source_split', 'most_impressing_split',
              'payment_mode_split', 'country_split', 'age_group_split']

for col in split_cols:
    train_data[col] = train_data[col].map({'high': 1, 'low': 0})
````

#### From Multivariate analysis

````python
# 12. Pupose of Leisure and Holidays spend more
train_data['is_leisure'] = (
    train_data['purpose'].isin(['Leisure and Holidays'])
).astype(int)
````

````python
# 13. Age x trave_with
# For Tourists in age groups 1-24, 25-44 and 45-64, they spen more if they are travelling with spouse and children or spouse alone.
# For Tourists who are ages 65+, they spend more if they are traveling with children alone or spouse alone
train_data['age_travel_interaction'] = (
    (train_data['age_group'].isin(['1-24', '25-44', '45-64']) & train_data['travel_with'].isin(['Spouse and Children', 'Spouse']) ) |
    ((train_data['age_group'] == '65+') & train_data['travel_with'].isin(['Spouse', 'Children']) )
).astype(int)
````

````python
# 14. age x main_activity
# For all ages, tourists who come for Conference tourism spend more
train_data['is_conference'] = (train_data['main_activity'] == 'Conference tourism').astype(int)

# For age groups 45-64 and 65+, tourists who come for Conference tourism and wildlife tourism spend more
train_data['age_conference_wildlife'] = (
    (train_data['age_group'].isin(['45-64', '65+'])) & (train_data['main_activity'].isin(['Conference tourism', 'Wildlife tourism'])) 
).astype(int)
````

````python
# 15. purpose x travel with
# For tourists whose purpose was leisure and holiday or meetings and conference, spend more when they travel with children
train_data['purpose_children_interaction'] = (
    (train_data['purpose'].isin(['Leisure and Holidays', 'Meetings and Conference'])) &
    (train_data['travel_with'] == 'Children')
).astype(int)

# For the tourists whose purpose was scientific and academic, they spend more when they travel with spouse
train_data['academic_spouse'] = (
    (train_data['purpose'] == 'Scientific and Academic') &
    (train_data['travel_with'] == 'Spouse')
).astype(int)
````

````python
# 16. payment_mode x travel with
# For tourists traveling with spouse and children or spouse alone, spend more if their payment mode is credit card or cash
train_data['payment_travel_interaction'] = (
    (train_data['travel_with'].isin(['Spouse and Children', 'Spouse'])) &
    (train_data['payment_mode'].isin(['Credit Card', 'Cash']) )
).astype(int)
````

````python
# 17. first trip x travel with
# For tourists who it's their first trip to TZ, they tend to spend more if they are travelling with spouse and children or chilren alone or spouse alone
train_data['first_trip_family'] = (
    (train_data['first_trip_tz'] == 1 ) &
    (train_data['travel_with'].isin(['Children', 'Spouse', 'Spouse and Children']) )
).astype(int)
````

````python
# 18. age x payment_mode
# Tourists in age group 65+ spend more if their payment mode is cash or credit card or other
train_data['senior_payment_high'] = (
    (train_data['age_group'] == '65+' ) &
    (train_data['payment_mode'].isin(['Cash', 'Credit Card', 'Other']) )
).astype(int)
````

````python
# 19. purpose x source of information
# Tourists who came for Leisure and holiday scientific and academic purpose and their source of information was inflight magazines spent more
train_data['inflight_high_spend'] = (
    (train_data['purpose'].isin(['Leisure and Holidays', 'Scientific and Academic']) ) &
    (train_data['info_source'] == 'inflight magazines' )
).astype(int)

# Tourists who came for volunteering, visiting friends and relatives and leisure and holidays purpose and their source of information was travel agent, tour operator spent more also
train_data['operator_high_spend'] = (
    (train_data['purpose'].isin(['Leisure and Holidays', 'Volunteering', 'Visiting Friends and Relatives']) ) &
    (train_data['info_source'] == 'Travel, agent, tour operator' )
).astype(int)
````

````python
# 20. activity x first_tz
# Tourists who it's their first time in TZ they spend more in Wildlife, beach, conference tourism and diving and sport fishing
train_data['activity_first'] = (
    (train_data['first_trip_tz'] == 1 ) &
    (train_data['main_activity'].isin(['Wildlife tourism', 'Conference tourism', 'Beach tourism', 'Diving and Sport Fishing']) )
).astype(int)

# Tourists spend more on conference tourism
train_data['is_conference'] = (
    train_data['main_activity'] == 'Conference tourism'
).astype(int)
````

````python
# 21. Num columns
train_data['total_people'] = train_data['total_male'] + train_data['total_female']
train_data['total_nights'] = train_data['night_mainland'] + train_data['night_zanzibar']

# 22. Tourists who do not stay the night in Zanzibar spend least
train_data['did_not_visit_zanzibar'] = (
    train_data['night_zanzibar_bin'] == 'none'
).astype(int)

# 24. or more male tourist who spend long or don't in mainland TZ they tend to spend more
train_data['male_mainland_tz'] = (
    (train_data['total_male'] > 0 ) &
    (train_data['night_mainland_bin'].isin(['long', 'none']) )
).astype(int)
````

````python
train_data.columns
````

````python
# 22. Encode the remaining feature for model creation
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
# Onehot ecoding for remaining features with many categories
encode_cols = ['country', 'travel_with', 'purpose', 'main_activity', 'info_source', 'payment_mode', 'most_impressing']

# Fit encoder on train
train_cat = ohe.fit_transform(train_data[encode_cols])
ohe_cols = ohe.get_feature_names_out(encode_cols)
# convert to df
train_cat = pd.DataFrame(train_cat, columns=ohe_cols, index=train_data.index)
# Drop original categorical columns
train_data = train_data.drop(columns=encode_cols)
# concatenate the encoded features
train_data = pd.concat([train_data, train_cat], axis=1)

# Ordinal encoding for age group, night bins
age_group_map = {
    '1-24': 1,
    '25-44': 2,
    '45-64': 3,
    '65+': 4
}
train_data['age_group'] = train_data['age_group'].map(age_group_map)

bin_map = {
    'none': 0,
    'short': 1,
    'medium': 2,
    'long': 3,
    'extended': 4
}

train_data['night_mainland_bin'] = (
    train_data['night_mainland_bin'].map(bin_map)
)

zanzibar_map = {
    'none': 0,
    'short': 1,
    'medium': 2,
    'long': 3
}

train_data['night_zanzibar_bin'] = (
    train_data['night_zanzibar_bin'].map(zanzibar_map)
)

train_data['night_mainland_bin'] = train_data['night_mainland_bin'].astype(int)
train_data['night_zanzibar_bin'] = train_data['night_zanzibar_bin'].astype(int)
````

````python
# Encode the remaining feature for model creation

# Onehot ecoding for remaining features with many categories
encode_cols = ['country', 'travel_with', 'purpose', 'main_activity', 'info_source', 'payment_mode', 'most_impressing']
# Fit Encoder on train
test_cat = ohe.transform(test_data[encode_cols])
# convert to df
test_cat = pd.DataFrame(test_cat, columns=ohe_cols, index=test_data.index)
# Drop original categorical columns
test_data = test_data.drop(columns=encode_cols)
# concatenate the encoded features
test_data = pd.concat([test_data, test_cat], axis=1)

# Ordinal encoding for age group
age_group_map = {
    '1-24': 1,
    '25-44': 2,
    '45-64': 3,
    '65+': 4
}
test_data['age_group'] = test_data['age_group'].map(age_group_map)

bin_map = {
    'none': 0,
    'short': 1,
    'medium': 2,
    'long': 3,
    'extended': 4
}

test_data['night_mainland_bin'] = (
    test_data['night_mainland_bin'].map(bin_map)
)

zanzibar_map = {
    'none': 0,
    'short': 1,
    'medium': 2,
    'long': 3
}

test_data['night_zanzibar_bin'] = (
    test_data['night_zanzibar_bin'].map(zanzibar_map)
)

test_data['night_mainland_bin'] = test_data['night_mainland_bin'].astype(int)
test_data['night_zanzibar_bin'] = test_data['night_zanzibar_bin'].astype(int)
````
